```python
import kagglehub
import os

path = kagglehub.dataset_download("sudinrupakheti/potholes")

# Update data.yaml with correct path
yaml_content = f"""path: {path}
train: images/train
val: images/val
test: images/test
nc: 1
names: ['pothole']
"""

with open("data.yaml", "w") as f:
    f.write(yaml_content)

# Install + train
os.system("pip install -q ultralytics")
os.system("yolo detect train data=data.yaml model=yolov8n.pt epochs=50")

# Copy weights
import shutil
shutil.copy("runs/detect/train/weights/best.pt", "/kaggle/working/best.pt")
print("✓ Done.")
```
